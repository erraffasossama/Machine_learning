```python
import numpy as np

class DBSCAN:

    def __init__(self, eps=1.5, minPts=3):
        self.eps = eps
        self.minPts = minPts

    # Distance euclidienne
    def distance(self, p1, p2):
        return np.sqrt(np.sum((p1 - p2) ** 2))

    # Recherche des voisins
    def region_query(self, X, point):

        neighbors = []

        for i in range(len(X)):
            if self.distance(X[point], X[i]) <= self.eps:
                neighbors.append(i)

        return neighbors

    # Expansion du cluster
    def expand_cluster(self, X, labels, point, neighbors, cluster_id):

        labels[point] = cluster_id

        i = 0

        while i < len(neighbors):

            neighbor = neighbors[i]

            if labels[neighbor] == -1:
                labels[neighbor] = cluster_id

            elif labels[neighbor] == 0:

                labels[neighbor] = cluster_id

                new_neighbors = self.region_query(X, neighbor)

                if len(new_neighbors) >= self.minPts:
                    neighbors.extend(new_neighbors)

            i += 1

    # Entraînement
    def fit(self, X):

        labels = np.zeros(len(X))

        cluster_id = 0

        for point in range(len(X)):

            if labels[point] != 0:
                continue

            neighbors = self.region_query(X, point)

            if len(neighbors) < self.minPts:

                labels[point] = -1

            else:

                cluster_id += 1

                self.expand_cluster(
                    X,
                    labels,
                    point,
                    neighbors,
                    cluster_id
                )

        return labels
```

```python
import numpy as np

X = np.array([
    [1,2],
    [2,2],
    [2,3],
    [8,8],
    [8,9],
    [25,25]
])

dbscan = DBSCAN(eps=2, minPts=2)

labels = dbscan.fit(X)

print(labels)
```
