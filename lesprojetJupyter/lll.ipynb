```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

class RegressionLineaire:
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.theta = None

    def fit(self, X, y):
        X_b = np.c_[np.ones((X.shape[0], 1)), X]
        self.theta = np.random.randn(X_b.shape[1], 1)
        for _ in range(self.n_iterations):
            gradients = 2/X_b.shape[0] * X_b.T.dot(X_b.dot(self.theta) - y.reshape(-1, 1))
            self.theta = self.theta - self.learning_rate * gradients

    def predict(self, X):
        X_b = np.c_[np.ones((X.shape[0], 1)), X]
        return X_b.dot(self.theta).flatten()

def calculer_r2(y_true, y_pred):
    SSE = np.sum((y_true - y_pred)**2)
    SST = np.sum((y_true - np.mean(y_true))**2)
    return 1 - (SSE / SST)

data = pd.read_csv('Salary_Data.csv')
X = data[['YearsExperience']].values
y = data['Salary'].values

plt.scatter(X, y)
plt.xlabel('YearsExperience')
plt.ylabel('Salary')
plt.show()

slr = RegressionLineaire()
slr.fit(X, y)
y_pred = slr.predict(X)

train_size = int(0.8 * len(X))
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]

y_test_pred = slr.predict(X_test)
MSE = np.mean((y_test - y_test_pred)**2)
R2_test = calculer_r2(y_test, y_test_pred)

print("MSE :", MSE)
print("R² test :", R2_test)

plt.scatter(X, y, color='blue', label='Data')
plt.plot(X, slr.predict(X), color='red', label='Regression lineaire')
plt.title("Regression lineaire simple")
plt.xlabel("YearsExperience")
plt.ylabel("Salary")
plt.legend()
plt.show()
```
